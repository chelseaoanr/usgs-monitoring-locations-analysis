```python
headerlist = []
datalist = []

# Start with the first page in JSON format
url = "https://api.waterdata.usgs.gov/ogcapi/v1/collections/monitoring-locations/items?limit=2000"

for page in range(25):

    # Request JSON so we can get the next-page link
    response = requests.get(url)
    data = response.json()

    # Convert the same page to HTML for BeautifulSoup
    html_url = url + "&f=html"
    html_response = requests.get(html_url)
    soup = BeautifulSoup(html_response.text, "html.parser")

    divs = soup.find_all("div", class_="usa-table-container")

    for div in divs:

        for table in div.find_all(
            "table",
            class_="usa-table usa-table--striped"
        ):

            # Collect column headers only once
            if len(headerlist) == 0:

                for headrow in table.find_all("thead"):
                    headers = headrow.find_all("tr")

                    for header in headers:
                        head = header.find_all("th")

                        for h in head:
                            headerlist.append(h.text.strip())

            # Collect data
            for datarow in table.find_all("tbody"):
                rows = datarow.find_all("tr")

                for row in rows:
                    data_cells = row.find_all("td")
                    rowlist = []

                    for d in data_cells:
                        rowlist.append(d.text.strip())

                    datalist.append(rowlist)

    print(
        "Page", page + 1,
        "complete - Total records:",
        len(datalist)
    )

    # Get next page URL from JSON
    next_url = next(
        (
            link["href"]
            for link in data.get("links", [])
            if link.get("rel") == "next"
        ),
        None
    )

    # Stop if there is no next page
    if next_url is None:
        print("No next page found.")
        break

    url = next_url

    time.sleep(0.5)
```

## 4.) Statistical Analysis

###     I.) How are USGS monitoring locations distributed across different states?

Frequency:

```python
state_frequency = df["state_name"].value_counts()
state_frequency
```

Percentage:

```python
state_percentage = df["state_name"].value_counts(normalize=True) * 100
state_percentage.round(2)
```

Mode:

```python
df["state_name"].mode()
```

Of the 50,000 USGS monitoring locations scraped, Arkansas contained 49,694 locations, representing approximately 99.39% of the dataset. Alabama contained 306 locations, representing approximately 0.61%. Arkansas was therefore the mode, meaning it was the most frequently occurring state in the scraped dataset.

### II.) How does the distribution of monitoring-location altitude differ among states?

```python
altitude_by_state = df.groupby("state_name")["altitude"].agg(
    Mean="mean",
    Median="median",
    Minimum="min",
    Maximum="max"
)

altitude_by_state
```

For Arkansas, there are valid numeric altitude values, so Python calculated:
- Mean: 206.88
- Median: 204
- Minimum: 0
- Maximum: 1600
  
For Alabama, every result is NaN. Meaning the 306 Alabama records do not have usable altitude values. 

```python
df.groupby("state_name")["altitude"].count()
```

Alabama can't be included in the altitude comparison because it has 0 valid altitude values, while Arkansas has 12,379 valid altitude measurements

```python
altitude_by_state["Q1"] = df.groupby("state_name")["altitude"].quantile(0.25)
altitude_by_state["Q3"] = df.groupby("state_name")["altitude"].quantile(0.75)
altitude_by_state["IQR"] = altitude_by_state["Q3"] - altitude_by_state["Q1"]

altitude_by_state.round(2)
```

Final Interpretation:

Alabama had no valid altitude values in the scraped dataset, so altitude statistics couldn't be calculated for that state. Arkansas had 12,379 valid altitude measurements. The mean altitude was 206.88, and the median was 204.00. Altitude values ranged from 0 to 1,600. The first quartile was 187.00 and the third quartile was 220.00, giving an IQR of 33.00. This means the middle 50% of Arkansas monitoring-location altitudes fell within a relatively narrow 33-unit range.

### III.) How much does altitude vary among the monitoring locations? 

```python
altitude = df["altitude"].dropna()

altitude_range = altitude.max() - altitude.min()
altitude_std = altitude.std()
altitude_variance = altitude.var()

Q1 = altitude.quantile(0.25)
Q3 = altitude.quantile(0.75)
altitude_iqr = Q3 - Q1
```

```python
print("Range:", round(altitude_range, 2))
print("Standard Deviation:", round(altitude_std, 2))
print("Variance:", round(altitude_variance, 2))
print("IQR:", round(altitude_iqr, 2))
```

The altitude measurements show a wide overall spread. The range is 1,600, the standard deviation is 53.07, the variance is 2,816.66, and the IQR is 33.0.
